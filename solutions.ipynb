## Задание 1: Линейный график «Температура за неделю»
**Условие:**
Постройте линейный график на основе данных о температуре за 7 дней.

**Данные:**
```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

**Требования:**
- Постройте график с помощью `plt.plot()`
- По оси X покажите дни, по оси Y - температуру.
- Добавьте заголовок графика.
- Используйте `marker` и измените стиль линии.
- Добавьте сетку (`grid`).
- Над каждой точкой покажите значение температуры.
- Измените размер графика с помощью `figsize`.

````python
# 1. Исходные данные
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]

# 2. Создание фигуры с заданным размером figsize
fig, ax = plt.subplots(figsize=(10, 6))

# 3. Построение графика: линия пунктиром, маркеры кругов, стилизация
ax.plot(
    days,
    temperature,
    color="#2b5c8f",
    linestyle="--",
    linewidth=2.5,
    marker="o",
    markersize=9,
    markerfacecolor="#ff7f0e",
    markeredgecolor="#2b5c8f",
    markeredgewidth=2,
    label="Дневная температура"
)

# 4. Подписи осей и заголовок
ax.set_title("Динамика температуры за неделю", fontsize=15, fontweight="bold", pad=15)
ax.set_xlabel("Дни недели", fontsize=12, labelpad=10)
ax.set_ylabel("Температура (°C)", fontsize=12, labelpad=10)
ax.set_ylim(min(temperature) - 3, max(temperature) + 3)
ax.grid(True, linestyle=":", alpha=0.6, color="gray")

# 5. Значения температуры над каждой точкой
for day, temp in zip(days, temperature):
    ax.text(
        x=day,
        y=temp + 0.45,
        s=f"{temp}°C",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color="#1f3b5c"
    )

ax.legend(loc="upper left", frameon=True)
plt.show()
````

## Задание 2: Столбчатые диаграммы «Сравнение языков программирования»
**Условие:**
Покажите количество студентов по языкам программирования на горизонтальной столбчатой диаграмме.

**Данные:**
```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
```

**Требования:**
- Создайте горизонтальную столбчатую диаграмму с помощью `barh()`
- Используйте отдельный цвет для каждого столбца.
- Подпишите оси X и Y и добавьте заголовок.
- В конце каждого столбца покажите количество студентов с помощью `plt.text()`
- Визуально выделите язык с наибольшим числом студентов.
- Измените размер графика с помощью `figsize`.

**Дополнительно:**
- Также покажите эти данные на отдельной вертикальной столбчатой диаграмме с помощью `plt.bar()`

````python
# 1. Исходные данные
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

# Для удобного чтения разворачиваем порядок, чтобы лидер был сверху
langs_h = languages[::-1]
studs_h = students[::-1]
max_val = max(studs_h)

# Цвета столбцов: лидер выделен оранжевым, остальные синим
colors_h = ["#ff9f1c" if val == max_val else "#4361ee" for val in studs_h]
edge_colors = ["#b25900" if val == max_val else "none" for val in studs_h]
line_widths = [2.5 if val == max_val else 0 for val in studs_h]

fig_h, ax_h = plt.subplots(figsize=(10, 6))

bars = ax_h.barh(
    langs_h,
    studs_h,
    color=colors_h,
    edgecolor=edge_colors,
    linewidth=line_widths,
    height=0.65
)

ax_h.set_title("Популярность языков программирования (горизонтальная диаграмма barh)", fontsize=14, fontweight="bold", pad=15)
ax_h.set_xlabel("Количество студентов", fontsize=12, labelpad=10)
ax_h.set_ylabel("Языки программирования", fontsize=12, labelpad=10)
ax_h.set_xlim(0, max_val + 7)
ax_h.grid(axis="x", linestyle="--", alpha=0.5)

# Подписи значений в конце каждого столбца
for i, val in enumerate(studs_h):
    is_leader = (val == max_val)
    label_text = f"{val} студ. (Лидер)" if is_leader else f"{val}"
    ax_h.text(
        x=val + 0.6,
        y=i,
        s=label_text,
        va="center",
        fontsize=10,
        fontweight="bold" if is_leader else "normal",
        color="#b25900" if is_leader else "#222222"
    )

plt.show()
````

````python
# ДОПОЛНИТЕЛЬНО: Вертикальная столбчатая диаграмма (plt.bar)
fig_v, ax_v = plt.subplots(figsize=(10, 6))
colors_v = ["#ff9f1c", "#4361ee", "#3a0ca3", "#7209b7", "#f72585"]

bars_v = ax_v.bar(
    languages,
    students,
    color=colors_v,
    edgecolor="black",
    linewidth=1.2,
    width=0.55
)

# Выделение столбца Python
bars_v[0].set_linewidth(2.8)
bars_v[0].set_edgecolor("#b25900")

ax_v.set_title("Сравнение языков программирования (вертикальная диаграмма bar)", fontsize=14, fontweight="bold", pad=15)
ax_v.set_xlabel("Языки программирования", fontsize=12, labelpad=10)
ax_v.set_ylabel("Количество студентов", fontsize=12, labelpad=10)
ax_v.set_ylim(0, max(students) + 6)
ax_v.grid(axis="y", linestyle="--", alpha=0.5)

for i, val in enumerate(students):
    ax_v.text(i, val + 0.8, str(val), ha="center", va="bottom", fontsize=11, fontweight="bold")

plt.show()
````

## Задание 3: Диаграмма рассеяния «Анализ результатов студентов»
**Условие:**
Проанализируйте часы учёбы, результаты экзамена и возраст на диаграмме рассеяния.

**Данные:**
```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

**Требования:**
- Используйте `fig = plt.figure()` и `fig.add_axes(...)`
- Создайте диаграмму рассеяния для `study_hours` и `scores`.
- Цвет точек должен зависеть от `ages`; используйте `cmap`.
- Используйте параметры `marker` и `s`.
- Добавьте `colorbar` с подписью `Age`.
- Подпишите ось X: `Study Hours`, ось Y: `Exam Score`.
- Добавьте сетку и подходящий заголовок.
- Отдельно отметьте точку с наивысшим результатом.

**Дополнительно:**
- В той же Figure создайте вторую небольшую область `Axes` и покажите данные `scores` в виде столбчатой диаграммы.

````python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]

# 1. Создание Figure
fig = plt.figure(figsize=(11, 7))

# 2. Основная область Axes через add_axes: [left, bottom, width, height]
ax_main = fig.add_axes((0.1, 0.12, 0.75, 0.78))

# 3. Диаграмма рассеяния
scatter = ax_main.scatter(
    study_hours,
    scores,
    c=ages,
    cmap="plasma",
    s=140,
    marker="o",
    edgecolors="black",
    linewidths=1.2,
    alpha=0.9
)

# 4. Colorbar с подписью Age
cbar = fig.colorbar(scatter, ax=ax_main, fraction=0.046, pad=0.04)
cbar.set_label("Возраст студента (Age)", fontsize=11, labelpad=10)

# 5. Подписи и оформление
ax_main.set_title("Зависимость балла на экзамене от часов подготовки", fontsize=14, fontweight="bold", pad=15)
ax_main.set_xlabel("Study Hours (Часы подготовки)", fontsize=12, labelpad=10)
ax_main.set_ylabel("Exam Score (Балл экзамена)", fontsize=12, labelpad=10)
ax_main.set_xlim(0, 9)
ax_main.set_ylim(40, 102)
ax_main.grid(True, linestyle="--", alpha=0.5)

# 6. Отметка точки с наивысшим результатом
max_score = max(scores)
max_idx = scores.index(max_score)
top_x, top_y = study_hours[max_idx], scores[max_idx]

ax_main.scatter(top_x, top_y, color="#ffd700", s=280, marker="*", edgecolors="black", linewidths=1.5, zorder=5, label=f"Максимум: {max_score} баллов")
ax_main.annotate(
    f"Наивысший балл!\nScore: {max_score}, Hours: {top_x}",
    xy=(top_x, top_y),
    xytext=(top_x - 3.2, top_y - 2),
    arrowprops=dict(facecolor="black", shrink=0.08, width=1.5, headwidth=8),
    fontsize=10,
    fontweight="bold",
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#fff3cd", edgecolor="#ffeeba", alpha=0.9)
)
ax_main.legend(loc="lower right", frameon=True)

# 7. ДОПОЛНИТЕЛЬНО: Вторая область Axes (вложенная диаграмма распределения баллов)
ax_inset = fig.add_axes((0.17, 0.58, 0.28, 0.26))
student_ids = [f"S{i+1}" for i in range(len(scores))]
bar_colors = ["#4cc9f0" if s != max_score else "#ffd700" for s in scores]
ax_inset.bar(student_ids, scores, color=bar_colors, edgecolor="black", linewidth=0.8)
ax_inset.set_title("Баллы по студентам (Scores)", fontsize=9, fontweight="bold")
ax_inset.set_xlabel("Студент", fontsize=8)
ax_inset.set_ylabel("Балл", fontsize=8)
ax_inset.tick_params(axis="both", labelsize=7)
ax_inset.set_ylim(0, 110)
ax_inset.grid(axis="y", linestyle=":", alpha=0.5)

plt.show()
````
